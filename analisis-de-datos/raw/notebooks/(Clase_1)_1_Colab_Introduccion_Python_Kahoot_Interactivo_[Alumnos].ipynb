# BLOQUE 3 — Strings

# BLOQUE 4 — Números y operadores

Python permite realizar **operaciones matemáticas directamente con valores numéricos**. Para hacerlo utiliza operadores, es decir, símbolos que indican qué operación queremos realizar.

# BLOQUE 5 — Booleanos y comparaciones

Los valores **booleanos** (`bool`) representan únicamente dos posibilidades: `True` (verdadero) o `False` (falso). En Python suelen aparecer como resultado de realizar una comparación entre valores.

Para comparar podemos utilizar operadores como `>`, `<`, `>=`, `<=`, `==` (igual) y `!=` (distinto).

También podemos **combinar varias condiciones** utilizando operadores lógicos. `and` devuelve `True` cuando ambas condiciones son verdaderas, mientras que `or` devuelve `True` cuando al menos una de ellas es verdadera.

En el siguiente ejercicio van a realizar distintas comparaciones y combinar algunas de ellas para **observar cuándo Python devuelve `True` y cuándo devuelve `False`**.

```python
edad = 20
nota = 7

resultado_1 = edad >= 18
resultado_2 = nota == 10
resultado_3 = edad < 20
resultado_4 = nota != 7

print(resultado_1)
print(resultado_2)
print(resultado_3)
print(resultado_4)
```

```python
print(resultado_1 and resultado_2)
print(resultado_1 or resultado_2)
```

Observen la secuencia completa de resultados.

# BLOQUE 6 — Listas

Una **lista** es una estructura que permite **guardar varios elementos dentro de una misma variable**. Estos elementos se almacenan en un orden determinado y cada uno ocupa una posición llamada **índice**.

En Python, los índices comienzan en `0`: el primer elemento tiene índice `0`, el segundo índice `1`, el tercero índice `2`, y así sucesivamente. También podemos acceder desde el final utilizando índices negativos: `-1` representa el **último elemento**, `-2` el anteúltimo, etc.

Por ejemplo, en la lista `["A", "B", "C"]`, `"A"` está en la posición `0`, `"B"` en la `1` y `"C"` puede encontrarse tanto con el índice `2` como con `-1`.

Además, podemos **modificar una lista**, agregando elementos, eliminándolos u ordenándolos.

En el siguiente ejercicio van a realizar distintas operaciones para observar **cómo acceder a sus elementos y cómo cambia su contenido**.

```python
numeros = [8, 3, 6]
print(numeros)
```

```python
numeros.append(10)
print(numeros)
```

```python
numeros.remove(3)
print(numeros)
```

```python
numeros.sort()
print(numeros)
```

El último `print()` muestra el estado final de la lista.

# BLOQUE 7 — Diccionarios

Un **diccionario** es una estructura que permite guardar varios datos dentro de una misma variable, organizados en pares **clave : valor**.

La **clave** funciona como un identificador que nos permite acceder a un dato específico, mientras que el **valor** es la información asociada a esa clave.

Por ejemplo, en `{"nombre": "Lucía", "edad": 22}`, `"nombre"` y `"edad"` son las claves, mientras que `"Lucía"` y `22` son sus respectivos valores.

A diferencia de las listas, donde accedemos a los elementos mediante su **índice**, en los diccionarios accedemos a la información utilizando sus **claves**.

También podemos **agregar nuevos pares clave-valor, modificar valores existentes o eliminar información**.

En el siguiente ejercicio van a realizar estas operaciones y observar cómo va cambiando el contenido del diccionario.

```python
persona = {"nombre": "Lucía", "edad": 22, "ciudad": "Córdoba"}
print(persona["nombre"])
print(persona["ciudad"])
```

```python
persona["edad"] = 23
persona["curso"] = "Python"
print(persona)
```

```python
persona.pop("ciudad")
print(persona)
```

Observen qué claves y valores quedaron al final.

# BLOQUE 8 — `print()` e `input()`

En Python, `print()` e `input()` permiten **mostrar y recibir información** durante la ejecución de un programa.

La función `print()` se utiliza para **mostrar información en pantalla**, como textos, valores de variables o resultados de operaciones.

La función `input()` permite **pedirle un dato al usuario**. El programa se detiene hasta que el usuario escribe una respuesta y presiona Enter.

Es importante tener en cuenta que `input()` devuelve la información ingresada como **texto (`str`)**, incluso cuando escribimos un número. Si necesitamos utilizar ese dato como un número, debemos convertirlo, por ejemplo, utilizando `int()` para números enteros.

En el siguiente ejercicio van a probar esta diferencia. Primero ingresen **Ana** como nombre y **20** como edad, y observen qué ocurre con el tipo de dato de `edad`.

# BLOQUE 9 — Condicionales: `if` y `else`

Los **condicionales** permiten que un programa **tome decisiones** y ejecute diferentes instrucciones dependiendo de si una condición se cumple o no.

La instrucción `if` evalúa una condición. Si esa condición es `True`, se ejecuta el bloque de código asociado. Si es `False`, podemos utilizar `else` para indicar qué debe ejecutarse en caso contrario.

Por ejemplo, podemos evaluar si una persona es mayor de edad mediante `edad >= 18`: si la condición se cumple, el programa ejecutará una acción; si no se cumple, ejecutará otra.

Es importante observar la **indentación**, ya que las instrucciones que aparecen indentadas debajo de `if` o `else` son las que pertenecen a cada bloque.

# BLOQUE 10 — `if`, `elif` y `else`

Cuando necesitamos evaluar **más de dos posibilidades**, podemos combinar `if`, `elif` y `else`.

- `if` evalúa la **primera condición**.
- `elif` permite evaluar **otra condición** si la anterior no se cumplió. Podemos utilizar más de un `elif` si existen varias alternativas.
- `else` se ejecuta cuando **ninguna de las condiciones anteriores se cumple**.

Python evalúa las condiciones **en orden, de arriba hacia abajo**. Cuando encuentra una condición verdadera (`True`), ejecuta ese bloque y no continúa evaluando las alternativas siguientes.

En el siguiente ejercicio van a ejecutar el mismo programa con **tres notas diferentes** y observar qué bloque se ejecuta en cada caso.

image.png

```python
nota = 95
if nota >= 90:
    print("Excelente")
elif nota >= 70:
    print("Bueno")
else:
    print("Necesitas mejorar")
```

```python
nota = 80
if nota >= 90:
    print("Excelente")
elif nota >= 70:
    print("Bueno")
else:
    print("Necesitas mejorar")
```

```python
nota = 60
if nota >= 90:
    print("Excelente")
elif nota >= 70:
    print("Bueno")
else:
    print("Necesitas mejorar")
```
