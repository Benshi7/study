# BLOQUE 1 — Indentación

En Python, la **indentación** son los espacios que aparecen al comienzo de una línea y sirven para indicar qué instrucciones pertenecen a un mismo bloque de código.

Esto es especialmente importante en estructuras como `if`, `else` y funciones. Las instrucciones que forman parte del mismo bloque deben tener el mismo nivel de indentación.

Ejecuten el siguiente ejemplo y observen qué instrucciones pertenecen al `if` y cuál queda fuera.

```python
edad = 20

if edad >= 18:
    print("Mayor de edad")
    print("Puede ingresar")

print("Fin del programa")
```

Ahora ejecuten con `edad = 15`.

```python
edad = 15

if edad >= 18:
    print("Mayor de edad")
    print("Puede ingresar")

print("Fin del programa")
```

Comparen ambos resultados y observen qué líneas dejan de ejecutarse cuando la condición es falsa.

# BLOQUE 2 — Librerías e `import`

Una **librería** es un conjunto de herramientas ya creadas que podemos utilizar para realizar distintas tareas sin tener que programarlas desde cero.

Para utilizar una librería en nuestro código usamos `import`.

Python incluye varias librerías estándar. En este ejemplo vamos a importar `math`, que contiene herramientas para trabajar con operaciones matemáticas.

```python
import math

print(math.sqrt(81))
print(math.sqrt(144))
print(math.pi)
```

`math.sqrt()` calcula una raíz cuadrada y `math.pi` contiene el valor de π.

Ahora ejecuten:

# BLOQUE 3 — Pandas

`pandas` es una librería de Python especializada en **trabajar y analizar datos organizados en tablas**.

Una de sus estructuras principales es el **DataFrame**, que organiza la información en filas y columnas.

En el siguiente ejemplo vamos a importar `pandas`, crear una tabla sencilla y observar su contenido.

```python
import pandas as pd

datos = {
    "producto": ["A", "B", "C"],
    "ventas": [120, 180, 90]
}

tabla = pd.DataFrame(datos)

print(tabla)
```

Ahora accedan solamente a la columna `ventas`.

```python
print(tabla["ventas"])
```

Finalmente calculen el promedio de ventas.

```python
print(tabla["ventas"].mean())
```

Observen el valor obtenido.

# BLOQUE 4 — Funciones

Una **función** es un bloque de instrucciones creado para realizar una tarea específica. Una vez definida, podemos ejecutarla todas las veces que necesitemos.

Para crear una función usamos `def`. Una función también puede recibir valores llamados **argumentos** y puede devolver un resultado mediante `return`.

Ejecuten:

```python
def saludar(nombre):
    print("Hola", nombre)

saludar("Ana")
saludar("Juan")
```

Ahora prueben una función que devuelve un resultado.

```python
def sumar(a, b):
    resultado = a + b
    return resultado

resultado_final = sumar(7, 5)

print(resultado_final)
```

Observen qué valor devuelve la función `sumar()`.

# BLOQUE 5 — Tuplas y conjuntos

Además de las listas y los diccionarios, Python posee otras estructuras para guardar varios valores.

Una **tupla** (`tuple`) es una colección ordenada cuyos elementos no se modifican una vez creada.

Un **conjunto** (`set`) guarda elementos únicos: si un valor aparece repetido, Python conserva una sola aparición.

Ejecuten:

```python
coordenadas = (40.0, -73.0)

print(coordenadas)
print(coordenadas[0])
print(coordenadas[1])
```

Ahora observen qué sucede con los valores repetidos en un conjunto.

```python
numeros = {1, 2, 2, 3, 3, 3}

print(numeros)
```

# BLOQUE 6 — Más operaciones con strings

En el primer notebook ya trabajaron con algunas operaciones sobre texto. Ahora van a probar otras herramientas que permiten **buscar, contar, comprobar y dividir strings**.

- `find()` busca la posición de un texto.
- `count()` cuenta cuántas veces aparece.
- `startswith()` y `endswith()` comprueban cómo comienza o termina.
- `split()` divide un string en varias partes.

Ejecuten:

```python
texto = "hola python, hola datos"

print(texto.find("python"))
print(texto.count("hola"))
print(texto.startswith("hola"))
print(texto.endswith("datos"))
print(texto.split(", "))
```

Observen especialmente el resultado de `split()` y la cantidad obtenida con `count()`.

# BLOQUE 8 — Floats: precisión y métodos

Los `float` representan números con parte decimal. Algunos valores decimales no pueden representarse de forma completamente exacta en la computadora, por eso a veces aparecen pequeñas diferencias.

Ejecuten:

```python
resultado = 0.1 + 0.2

print(resultado)
print(resultado == 0.3)
```

Podemos utilizar `round()` para redondear el resultado.

```python
print(round(resultado, 1))
```

Los floats también poseen algunos métodos. `is_integer()` permite comprobar si un número decimal representa exactamente un entero.

```python
a = 5.0
b = 5.2

print(a.is_integer())
print(b.is_integer())
```

Observen los resultados de ambas pruebas.

# BLOQUE 10 — Más operaciones con listas

En el primer notebook usaron `append()`, `remove()` y `sort()`. Ahora van a probar otras operaciones incluidas en el material.

- `extend()` agrega varios elementos.
- `insert()` agrega un elemento en una posición determinada.
- `pop()` elimina un elemento por su posición.
- `index()` devuelve la posición de un elemento.
- `count()` cuenta cuántas veces aparece.
- `reverse()` invierte el orden.

Ejecuten las celdas **en orden**.

```python
numeros = [10, 20, 30]

numeros.extend([40, 50])
print(numeros)
```

```python
numeros.insert(1, 15)
print(numeros)
```

```python
eliminado = numeros.pop(3)

print(eliminado)
print(numeros)
```

```python
print(numeros.index(40))
print(numeros.count(20))

numeros.reverse()
print(numeros)
```

Observen el contenido final de la lista y el valor que eliminó `pop()`.

# BLOQUE 11 — Copiar una lista

Podemos crear una copia de una lista utilizando `copy()`.

Esto permite trabajar con una nueva lista sin modificar directamente la original.

Ejecuten:

```python
original = [1, 2, 3]

copia = original.copy()

copia.append(4)

print("Original:", original)
print("Copia:", copia)
```

Observen si agregar `4` a la copia modificó o no la lista original.

# BLOQUE 12 — Más operaciones con diccionarios

Los diccionarios poseen métodos que permiten consultar su información:

- `keys()` devuelve las claves.
- `values()` devuelve los valores.
- `items()` devuelve los pares clave-valor.
- `get()` permite consultar un valor mediante una clave.

Ejecuten:

```python
persona = {
    "nombre": "Ana",
    "edad": 25,
    "ciudad": "Buenos Aires"
}

print(persona.keys())
print(persona.values())
print(persona.items())
```

Ahora prueben `get()`.

```python
print(persona.get("edad"))
print(persona.get("curso", "No informado"))
```

Observen qué devuelve Python cuando la clave existe y qué ocurre cuando no existe.
