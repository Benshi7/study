```python
import pandas as pd
from ydata_profiling import ProfileReport
```

```python
# Cada notebook es independiente: hay que volver a importar librerías y cargar los datos.
# Si aparece FileNotFoundError: el CSV no está subido en el panel "Archivos" de ESTE notebook.
df = pd.read_csv("sample_dataset.csv")
df.shape   # esperamos (569, 31)
```

## 2) Un primer reporte (modo mínimo)

📘 **Concepto.** `ProfileReport(df)` arma el reporte. Dos parámetros útiles:
- `minimal=True` → versión **liviana y rápida** (sin correlaciones ni interacciones). Ideal para datasets grandes o para una primera mirada.
- `title="..."` → título del reporte.

Empezamos con **solo 5 columnas** para que no sea tan complejo ni tan lento.

```python
#vamos a trabajar con solo 5 columnas del dataset para que no sea tan complejo
report = ProfileReport(df.iloc[:, 0:5], minimal=True)
```

```python
#Vamos a generar el reporte automático del Análisis Exploratorio con esta librería
report.to_notebook_iframe()
#Vemos un Overview de nuestro Dataset
```

📘 **Cómo se lee el reporte:**
- **Overview → Dataset statistics:** filas, variables, celdas faltantes (*Missing cells*), filas duplicadas, memoria, tipos de variables.
- **Overview → Alerts:** la lista de **alertas** (lo más valioso del reporte).
- **Variables:** una sección por variable, con estadísticos (*Distinct*, *Missing*, *Mean*, *Min*, *Max*, *Zeros*...) y un histograma. Con *More details* aparecen **cuantiles**, **IQR**, **asimetría** (*Skewness*) y más.

## 3) Un reporte más completo, con la variable objetivo

🧩 Ahora armamos un reporte **sin** el modo mínimo, para que incluya **correlaciones**, **interacciones** (scatterplots) y el análisis de **faltantes**.

## 4) 🚨 Interpretar las alertas

📘 Las alertas más comunes de ydata-profiling y su relación con la teoría (lámina *"Identificación de problemas en los datos"*):

| Alerta | Qué significa | Concepto de la teoría |
|---|---|---|
| `Missing` | la variable tiene valores faltantes | Valores faltantes (NaN) |
| `Duplicates` | hay filas repetidas | Registros duplicados |
| `High correlation` | dos variables están muy correlacionadas | Correlación / redundancia (notebook 5) |
| `Imbalance` | una categoría domina casi todos los valores | Variable casi constante (notebook 1) |
| `Skewed` | distribución muy asimétrica | Asimetría (notebook 2) |
| `Zeros` | muchos valores exactamente 0 | Posible valor inconsistente o "dato no medido" |
| `Constant` / `Unique` | un solo valor / todos distintos | Variable sin información / identificador (como un `id`) |
